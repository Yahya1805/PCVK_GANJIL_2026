```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt
```

```python
urls = [
    "https://picsum.photos/id/10/200/150",
    "https://picsum.photos/id/20/200/150",
    "https://picsum.photos/id/30/200/150"
]

for url in urls:
    image = io.imread(url)
    image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
    final_frame = cv.hconcat((image, image_2))
    cv2_imshow(final_frame)
    print('\n')
```

```python
# Membuat salinan gambar
img_garis = image_2.copy()

# Membuat garis vertikal berwarna putih
cv.line(img_garis, (50, 20), (50, 130), (255, 255, 255), 2)

# Membuat garis diagonal berwarna putih
cv.line(img_garis, (20, 20), (150, 130), (255, 255, 255), 2)

# Menampilkan gambar
cv2_imshow(img_garis)
```

```python
# Membuat salinan gambar
img_horizontal = image_2.copy()

# Mendapatkan ukuran gambar
height, width = img_horizontal.shape[:2]

# Membuat garis horizontal putih di tengah gambar
cv.line(
    img_horizontal,
    (20, height // 2),
    (width - 20, height // 2),
    (255, 255, 255),
    3
)

# Menampilkan gambar
cv2_imshow(img_horizontal)
```

```python
# Membuat salinan gambar
img_kotak = image_2.copy()

# Membuat kotak berwarna putih
cv.rectangle(
    img_kotak,
    (80, 40),
    (130, 90),
    (255, 255, 255),
    -1
)

# Menampilkan gambar
cv2_imshow(img_kotak)
```

```python
# Membuat salinan gambar
img_pixel = image_2.copy()

# Membuat kotak putih menggunakan kumpulan pixel
img_pixel[40:90, 80:130] = [255, 255, 255]

# Menampilkan gambar
cv2_imshow(img_pixel)
```
